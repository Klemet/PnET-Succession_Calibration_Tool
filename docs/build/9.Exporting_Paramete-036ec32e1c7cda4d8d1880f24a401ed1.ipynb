# Exporting the parameters to LANDIS-II files

*Most of the text in this notebook is directly from the [PnET-Succession Calibration Tool](https://github.com/Klemet/PnET-Succession_Calibration_Tool).*

Now that we have calibrated the parameters, it's time to export them into files that can be used directly in LANDIS-II.

In particular, we're going to generate :

- A core species parameter file
- A PnET-Succession species parameter file
- A PnET-Succession generic parmameter files

The core species and PnET-Succession species parameter file will be easy to do. The generic parameter file, however, will necessitate a couple of inputs, like the latitude of your landscape and the number of CPU cores you want to use to make the simulation faster (via the Parallel parameter). This later parameter depends on the computer or cluster you will use.

:::{tip} How to create your PnET-Succession ecoregion parameter file
In addition to its species parameter file and generic parameters file, PnET-Succession requires a main parameter file and an ecoregion parameter file. You can also give to it a "disturbance reduction file" so that it knows how different disturbances (fire, harvest, etc.) impact the different biomass pools of your age cohorts.

The ecoregion parameter file is going to be the more complex file to do. These calibration notebooks do not currently contain function to create it, as it will depend to how you define ecoregions in your landscape. The PnET-Succession ecoregion parameter file mostly contains informations about the soils of your ecoregions; mainly their texture, and the amount of fast leakage or runnof, etc. I refer to the PnET-Succession User Guide for you to generate this file.
:::

There are several optional generic parameter that I'm not using in these notebooks, but that you might want to use. Here are some examples (see the PnET-Succession user guide for the full list) : 

- MaxCanopyLayers : Limits the number of canopy layers PnET-Succession manages in each cell. Can be used to reduce the computations.
- LeakageFrostDepth : Influences the leakage where your soil are frozen.
- LifeForm : Specify the life form (stature) of the species to determine if there are limits to the species’ major canopy layer (tree, bush, etc.). By default, all species are trees. You can indicate this parameter as a species parameter to vary it per species (e.g. having trees and bushes in your simulation).
- CohortBinSize : Allows the lumping of age cohorts of different ages into a single cohort to reduce computations. Disabled by default.
- CO2AmaxBeff : Effect of the CO2 on AmaxB (efficiency of photosynthesis). Disabled by defaults.


```python
# Export the parameters into three files ready for LANDIS-II
from functionsForCalibration import *
import os
import shutil

dictOfCoreSpeciesParameters = json.load(open('./SpeciesParametersSets/Final_Parameters/coreSpeciesParameters.json'))
dictOfPnETSpeciesParameters = json.load(open('./SpeciesParametersSets/Final_Parameters/PnETSpeciesParameters.json'))
dictOfPnETGenericParameters = json.load(open('./SpeciesParametersSets/Final_Parameters/genericParameters.json'))


# Gathering the average latitude of the landscape from the shapefile used in notebook 0.2.
gdf = gpd.read_file(json.load(open('./ReferencesAndData/SpatialBoundaries/studyLandscapeInformation.json'))["studyLandscapeShapefilePath"])
# Check the CRS (Coordinate Reference System) of the shapefile
# If not in WGS84 (EPSG:4326), reproject it
if gdf.crs != 'EPSG:4326':
    gdf = gdf.to_crs('EPSG:4326')
# Get the latitude and put in the the generic parameters
center_lat = gdf.unary_union.centroid.y
dictOfPnETGenericParameters["Latitude"] = str(round(center_lat, 3))

# Indicating the number of CPU cores you want
# Indicate NONE to remove the generic parameter "Parallel", making sure your simulations won't use parallelisation.
dictOfPnETGenericParameters["Parallel"] = "4"

# Deciding the unit of PAR - Depend on what you want to use in your climate files.
dictOfPnETGenericParameters["PARunits"] = "umol"

# Setting prevent establishment to False
dictOfPnETGenericParameters["PreventEstablishment"] = "False"


exportFolder = "./SpeciesParametersSets/Final_Export/"
if os.path.exists(exportFolder):
    shutil.rmtree(exportFolder)
os.mkdir(exportFolder)

write_LANDIS_SpeciesCoreFile(exportFolder + "Core_Species_Parameters.txt", dictOfCoreSpeciesParameters)

write_LANDIS_SimpleParameterFile(exportFolder + "PnET_Generic_Parameters.txt", dictOfPnETGenericParameters)

write_PnET_ComplexTableParameterfile(exportFolder + "PnET_Species_Parameters.txt", dictOfPnETSpeciesParameters)
```
